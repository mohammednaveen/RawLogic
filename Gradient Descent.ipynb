# Gradient Descent from scratch

```python
# Dataset
x = [1, 2, 3, 4]
y = [3, 5, 7, 9]

#inititalizing values
w = 0.0
b = 0.0
lr = 0.01
n = len(x)
cost = 0
cost_vis = []
#training loop
for epochs in range(1000):

    dw = 0
    db = 0
    cost = 0
    for i in range(n):
        y_predicted = w * x[i] + b
        error = y_predicted - y[i]

        #accumulating partial derivatives to update the weight and bias
        dw += error * x[i]
        db += error
        cost += error ** 2

    cost = cost / n
    cost_vis.append(cost)
    
    dw = (2/n) * dw
    db = (2/n) * db

    w = w - lr * dw
    b = b - lr * db

#Final Parameters
print("weight :",w)
print("COst:",cost)
print("Bias :",b)     
```

```python
import matplotlib.pyplot as plt
plt.figure()
plt.plot(cost_vis)
plt.xlabel("Epochs")
plt.ylabel("Loss")
plt.title("Gradient Descent Visualization")
plt.show()
```

# Gradient Descent using Numpy

```python
import numpy as np
import matplotlib.pyplot as plt

# Dataset
x = np.array([5, 7, 9, 11, 13, 15], dtype=float)
y = np.array([15, 21, 27, 33, 39, 45], dtype=float)

# Parameters
w = 0.0
b = 0.0
lr = 0.001
epochs = 100
n = len(x)
loss = 0
losses = []

# Training loop
for epoch in range(epochs):

    y_pred = w * x + b
    error = y_pred - y

    loss = np.mean(error ** 2)
    losses.append(loss)

    dw = (2 / n) * np.sum(error * x)
    db = (2 / n) * np.sum(error)

    w = w - lr * dw
    b = b - lr * db

print("Final weight:", w)
print("Final bias:", b)
print("cost ",loss)
```

```python
plt.figure()
plt.plot(losses)
plt.xlabel("Epochs")
plt.ylabel("Loss (MSE)")
plt.title("Loss vs Epochs")
plt.show()
```
