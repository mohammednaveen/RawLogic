## Finding the Determinant of a Matrix

```python
def determinant(matrix):
    n = len(matrix)
    if any(len(row) != n for row in matrix):
        raise ValueError("Determinant is defined only for square matrices")
    if n == 1:
        return matrix[0][0]
    if n == 2:
        return matrix[0][0]*matrix[1][1] - matrix[0][1]*matrix[1][0]
    det = 0
    for col in range(n):
        minor = [
            row[:col]+row[col+1:]
            for row in matrix[1:]
        ]
        sign = (-1) ** col
        det += sign * matrix[0][col] * determinant(minor)
    return det
determinant(mat)
```

## Finding the Inverse of a Matrix

### Transpose function

```python
def transpose(matrix):
    return list(map(list,zip(*matrix)))
```

### Co-Factor Matrix function

```python
def get_cofactor(matrix):
    n = len(matrix)
    cofactor_matrix = []
    for i in range(n):
        row = []
        for j in range(n):
            minor = getMinor(matrix,i,j) # matrix after removing corresponding row and column
            sign = (-1) ** (i+j)
            row.append(sign*determinant(minor)) # finding cofactor
        cofactor_matrix.append(row)
    return cofactor_matrix
```

### Minor Function

```python
def getMinor(matrix,i,j): # i-th row and j-th column
    return [
        row[:j] + row[j+1:] # excluded j-th column
        for idx, row in enumerate(matrix) if idx != i # excluded i-th row
    ]
```

### All together Inverse

```python
def getInverse(matrix):
    n = len(matrix)
    det = determinant(matrix)
    if det == 0:
        raise ValueError("Determinant of matrix = 0, NO INVERSE!!")
    cofactor_matrix = get_cofactor(matrix)
    adj = transpose(cofactor_matrix)
    inverse = [
        [adj[i][j] / det for j in range(n)]
        for i in range(n)
    ]
    return inverse
```

### Using Gauss-Jordan Elimination - Computing Inverse 

```python
import numpy as np
def GaussJordanInverse(matrix):
    # I = np.eye(len(matrix))
    # matrix = np.hstack((matrix,I))
    n = len(matrix)
    #forward elimination
    for i in range(n): # make pivot one
        pivot = matrix[i][i]
        for j in range(n*2):
            matrix[i][j] /= pivot # dividing with the same number to make it one

        #make other columns in the matrix zero
        for k in range(n): # in each row
            if k != i: # every row other than the same row
                factor = matrix[k][i] # this is the value we want to eliminate (make 0)
                for j in range(2*n):
                    matrix[k][j] -= factor*matrix[i][j] #we already changed the pivot rows value that is (matrix[i][j]) as 1 so we can multiply that one with the value we want to eliminate and subtract it with the same value in the below(current) row to make it 0 (eliminate)

        # extract the inverse from the joined matrix
    inverse = [row[n:] for row in matrix]
    return inverse

aug = [
    [1, 2, 1, 0],
    [3, 4, 0, 1]
]
np.array(aug)
inv = GaussJordanInverse(aug)
print(inv)
```

## Solving Linear Equations using Matrices

### Using Gaussian Elimination

```python
def solve_linalg(matrix):
    n = len(matrix)
    # forward elimination
    for i in range(n): # for each row 
        for j in range(i+1,n): # forward rows
            factor = matrix[j][i]/matrix[i][i] # fiding by which value to reduce that row 
            for k in range(len(matrix[0])): # fetching each value in that row
                matrix[j][k] -= factor*matrix[i][k] # reducing each value by the multiplied version of the above row
    # after this we have reached the row echelon form
    # back subtitution
    x = [0]*n # vector for storing solutions
    for i in range(n-1,-1,-1): # starting from the back of the matrix
        x[i] = matrix[i][-1] / matrix[i][i] # taking the coefficient from one side to another and dividing to get the value for the solution
        # now we have to remove the found variable from the above equations and make that column 0 and the answer to the equation corredct( after the known 
        # value has been deducted from the equation we have to deduct that fromt the answer (augemted right part) too)
        # note: we are not changing the entire column of the known value to zero because we are not going to use that column anymore we will just traverse 
        # through the diagonal or pivot values and calculate the solution so instead, here we are only changing the answer(augmented right side) value only
        for j in range(i-1,-1,-1): 
            matrix[j][-1] -= matrix[j][i] * x[i]
    return x
solve_linalg([[2,1,5],[1,3,6]]) #we are passing augmented matrix
```

# Eigenvalues and Eigenvectors

### Finding Eigenvalues first

```python
#define matrix
A = [[4,1],
    [2,3]]
a,b = 4,1
c,d = 2,3

#coerfficients of quadratic equation
trace = a+d
determinant = a*d - b*c

#descriminant
disc = trace**2 - 4*determinant

#calculate eigenvalues
lambda1 = (trace - disc**0.5) / 2
lambda2 = (trace + disc**0.5) / 2

print("Eigenvalues :",lambda1,lambda2)
```

### Finding Eigenvectors 

```python
def eigenvectors(a,b,c,d,lam):
    #solve (a-lam)x + b*y = 0
    if b != 0:
        x = 1 # fix a value for x to calculate value of y 
        y = -(a - lam) / b
    if c != 0:
        y = 1
        x = -(d - lam) / c
    return (x,y)
print("EGV 1 :",eigenvectors(a,b,c,d,lambda1))
print("EGV 2 :",eigenvectors(a,b,c,d,lambda2))
```

### Covariance

```python
# Data 
X = [2,4,6,8]
Y = [1,3,5,7]

#Calculate the mean 
mean_X = sum(X)/len(X)
mean_Y = sum(Y)/len(Y)

# Calculater Variance of (X)
var_X = sum(x - mean_X for x in X) / (len(X) - 1)
#Calcaulate Variance of (Y)
var_Y = sum(y - mean_Y for y in Y) / (len(Y) - 1)
#Calculate Covariance of (X,Y)
cov_XY = sum((X[i] - mean_X)*(Y[i] - mean_Y) for i in range(len(X))) / (len(X)-1)

#Covariance Matrix
cov_matrix = [[var_X,cov_XY],[cov_XY,var_Y]]

print("Covariance Matrix (from scratch) :")
for row in cov_matrix:
    print(row)
```
