```python
# Just a function to restate the nodes to a specific set to perform operations :

# classs defining Node structure 
class ListNode:
    def __init__(self, val=0, next=None):
        self.val = val
        self.next = next

# node creation
def reestablish():
    node1 = ListNode(10)
    node2 = ListNode(20)
    node3 = ListNode(30)
    node4 = ListNode(40)
    node5 = ListNode(50)
    # Linking nodes to each other
    node1.next = node2 # head node = node1
    node2.next = node3
    node3.next = node4
    node4.next = node5

    head = node1
    return head
```

### Insertion at End/Tail

```python
# to find the end node we shall traverse till the end node which can be 
# identified using -> the current.next being None i.e the last node is pointing
# towards None
def InsertTail(head, new_node) :

    if head is None:
        return new_node

    current = head
    while current.next: # here is where the O(n) occurs
        current = current.next
    # now current is at the last node

    current.next = new_node
    return head

head = InsertTail(head, ListNode(50))

traversal()
```

### Linked List Reversal 

```python
def LL_reversal(head):
    prev = None
    current = head
    while current:
        next_node = current.next # save next before breaking the link
        current.next = prev # reverse pointer
        prev = current # Move Forward
        current = next_node
    return prev

head = LL_reversal(head)

traversal()
```

### Insert at Specified Index

```python
def InsertIndex(head, new_node, idx):
    if idx == 0:
        # new_node.next = head
        # head = new_node
        return InsertHead(head, new_node)
    
    else :
        current = head
        count = 0
        while current is not None and count < idx -1 :
            current = current.next # this will stop at current being the node before which insertion should happen : current - 30
            count += 1 
        if current is None:
            print("Index Not Found")
            return head
        new_node.next = current.next # joined new node to the rest of the list 
        current.next = new_node # joined 30 node to new node
        return head

head = InsertIndex(head, ListNode(80), 8)

traversal()
```

### Insert Node after a specified value

```python
def InsertValue(head, new_node, value):
    current = head 
    while current.val != value and current is not None:
        current = current.next
    if current is None:
        print('Value Not Found!!')
        return head
    new_node.next = current.next
    current.next = new_node
    return head

head = InsertValue(head,ListNode(70),30)

traversal()
```

### Delete Last Node

```python
def DeleteLast(head):
    current = head
    if current is None or current.next is None:
        print("Only one node - Node distroyed!!")
        return None
    while current.next.next:
        current = current.next
    current.next = None
    return head
head = DeleteLast(head)
traversal()
```

### Delete Specified Index

```python
def DeleteIndex(head, idx):
    if head is None or head.next is None:
        print('Only one node/ No node - Node destroyed!!')
        return None
    elif idx == 0:
        return head.next
    current = head
    count = 0
    while current is not None and count < idx-1:
        current = current.next
        count += 1
    if current is None or current.next is None:
        print("No such index exists!")
        return head
    current.next = current.next.next
    return head
```

### Delete A Node by Value

```python
def DeleteValue(head, value):
    while head and head.val == value:
        head = head.next
    current = head
    while current and current.next:
        if current.next.val == value:
            current.next = current.next.next
        else:
            current = current.next
    return head        
```

### Detecting A Cycle In The List

```python
# Floyd's Cycle Detection / Tortoise and Hare Algorithm
# since the fast goes 2 steps at a time it should meet slow at some point because 
def Cycle(head):
    slow = head
    fast = head
    while fast and fast.next:
        slow = slow.next
        fast = fast.next.next
        if slow == fast:
            return True
    return False
```

### Sorting A Linked List

```python
def sortList(head):
    if head is None or head.next is None:
        return head
    
    slow = fast = head
    prev = None # middle element
    while fast and fast.next:
        prev = slow # just beore the middle node
        slow = slow.next
        fast = fast.next.next
    # split into two halves
    prev.next = None # break the list
    # recursively sort both halves
    left = sortList(head)
    right = sortList(slow)

    return merge(left, right)
def merge(left, right):
    dummy = ListNode(0)
    tail = dummy

    while left and right:
        if left.val < right.val:
            tail.next = left
            left = left.next
        else:
            tail.next = right
            right = right.next
        tail = tail.next

    #Attach remaining nodes
    if left:
        tail.next = left
    if right:
        tail.next = right

    return dummy.next
```

### Remove duplicates From Sorted Linked List

```python
def DupeSorted(head):
    current = head
    while current and current.next:
        if current.val == current.next.val:
            current.next = current.next.next
        else:
            current = current.next
    return head
```

### Remove Duplicates from an Unsorted Linked List

```python
def DupeUnsorted(head):
    if head is None or head.next is None:
        return head
    current = head
    seen = set()
    prev = None
    while current:
        if current.val  not in seen:
            seen.add(current.val)
            prev = current # points at the first dupe
        else:
            prev.next = current.next # skips the dupe 
        current = current.next
    return head
```

### Finding nth Node From the End

```python
def FindFromEnd(head,n):
    slow = fast = head
    count = 0 
    for _ in range(n):
        if fast is None:
            return None
        fast = fast.next
    if fast is None:
        return head.val
    while fast :
        slow = slow.next
        fast = fast.next
    return slow.val
```

### Finding the Intersection Point of a Linked List:

```python
def IntersectionPoint(head1, head2):
    ptr1 = head1
    ptr2 = head2
    
    while ptr1 != ptr2:
        ptr1 = ptr1.next if ptr1 else head2
        ptr2 = ptr2.next if ptr2 else head1

    return ptr1
```

### Reverse a Linked List Recursively

```python
def reverseList(head):
    if not head:
        return None
    newHead = head
    if head.next:
        newHead = reverseList(head.next)
        head.next.next = head
        head.next = None
    return newHead
```

### Reversing K Groups

```python
def reverseKGroup(head, k):
    # create a dummy node
    dummy = ListNode(0)
    dummy.next = head
    group_prev = dummy

    while True:
        # Find the kth node if not then the list has no sufficient elements therefore return it
        kth = group_prev
        for _ in range(k):
            kth = kth.next
            if not kth:
                return dummy.next
        
        # store the kth node, we will join the reversed node to this node
        group_next = kth.next
        
        # reverse the group
        prev = group_next # joins it here
        curr = group_prev.next
        while curr != group_next:
            temp = curr.next
            curr.next = prev
            prev = curr
            curr = temp

        # Reconnect
        temp = group_prev.next # group_prev still points to dummy at the first so .next is first element 1
        # we need temp - [1] because it is the new tail of the reversed group
        group_prev.next = kth # dummy now shift to kth - [2]. i.e dummy -> [2] -> [1] -> [3]...
        group_prev = temp # we move dummy to next group [1]
```

### Check Palindrome

```python
def PalindromeList(head):
    if not head or not head.next:
        return True
    slow = fast = head
    while fast and fast.next:
        slow = slow.next
        fast= fast.next.next 
    prev = None
    curr = slow
    while curr:
        next_node = curr.next
        curr.next = prev
        prev = curr
        curr = next_node
    left = head
    right = prev
    while right:
        if left.val != right.val:
            return False
        left = left.next
        right = right.next
    return True
```

### Partition List around values

```python
def PartitionList(head,x):
    small_val = ListNode(0)
    large_val = ListNode(0)
    res = small_val
    temp = large_val
    current = head
    while current:
        if current.val < x:
            small_val.next = current
            small_val = small_val.next
        else:
            large_val.next = current
            large_val = large_val.next
        current = current.next
    large_val.next = None
    small_val.next = temp.next
    
    return res.next
```

### Circular Linked List
