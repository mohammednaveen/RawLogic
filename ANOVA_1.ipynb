# One-Way ANOVA  
## from scratch

Three different teaching methods are being tested to see if they produce different average exam scores among students.

Group A: taught using traditional lectures

Group B: taught using interactive videos

Group C: taught using AI-based personalized learning

Each group has 20 students, and their final exam scores (out of 100) are summarized as follows:

Group	Sample Size (n)	Mean Score (x̄)	Standard Deviation (s)

A	20	72	5

B	20	78	6

C	20	85	7

You need to determine whether there is a statistically significant difference in the mean exam scores among the three teaching methods.

Use a one-way ANOVA test at a 5% significance level (α = 0.05).

---

Tasks:

1. State the null and alternative hypotheses.
2. Compute the ANOVA table (SSB, SSW, MSB, MSW, F).
3. Determine the p-value or compare the computed F with the critical F-value.
4. State your conclusion in context — do the teaching methods lead to significantly different mean exam scores?

```python
# import necessary libraries
import math
from scipy import stats
```

Null Hypothesis : mean exam scores are equal among the three teaching techniques

Alternate Hypothesis : mean exam scores are not equal among the three teaching techniques

### Step 1
#### Define group summary data

```python
nA = nB = nC = 20
mean_A , mean_B , mean_C = 72,78,85
std_A , std_B , std_C = 5,6,7
N = nA+nB+nC # total sample size
k = 3 # Number of groups

print("Group sizes :",nA,nB,nC)
print("Group means :",mean_A,mean_B,mean_C)
print("Standard Deviation of groups :",std_A,std_B,std_C)
print("Total Sample size :",N)
print("Number of groups :",k)
```

### Step 2
#### Grand mean weighted by group sizes

```python
# Average of ALL numbers together.
grand_mean = ((nA * mean_A) + (nB * mean_B) + (nC * mean_C)) / N
print(f"Grand Mean : {grand_mean:.6f}")
```

### Step 3
#### Between-group SUM OF SQUARES (SSB)

```python
# How far each group mean is from grand mean. or how far are the group centers
SSB_A = nA * ((mean_A - grand_mean) ** 2)
SSB_B = nB * ((mean_B - grand_mean) ** 2)
SSB_C = nC * ((mean_C - grand_mean) ** 2)
SSB = SSB_A + SSB_B + SSB_C
print("SSB contributions :",SSB_A,SSB_B,SSB_C)
print(f"The Sum of Squares between groups (SSB) : {SSB:.6f}")
```

### Step 4
#### Within-group SUM OF SQUARES (SSW)

```python
# How scattered students are in each group or How much do the individual points wiggle around their own centers?
SSW_A = (nA - 1) * (std_A **2)
SSW_B = (nB - 1) * (std_B **2)
SSW_C = (nC - 1) * (std_C **2)
SSW = SSW_A + SSW_B + SSW_C
print("SSW contributions :",SSW_A,SSW_B,SSW_C)
print(f"The Sum of Squares Within-groups : {SSW:.6f}")
```

In statistics, when we calculate the variance of a sample to estimate a population, we lose one "degree of freedom."
- The Constraint: To calculate the standard deviation of Group A, you first have to calculate the mean of Group A.
- The Penalty: Once you know the mean, the first $n - 1$ numbers in your group can be anything, but the very last number must be a specific value for the mean to stay what it is. It's no longer "free" to vary.
- The Bias: Using just $n$ tends to underestimate the true spread of the population. Dividing by $n-1$ makes the number slightly larger, which gives us a more accurate, "unbiased" estimate of the variance.

### Step 5
#### Degrees of Freedom and Mean Squares 

```python
diff_between = k - 1
# This scales your SSB (Sum of Squares Between) so you are looking at the average variation between group centers, rather than just the total.
diff_within = N - k
MSB = SSB / diff_between 
MSW = SSW / diff_within
F_stat = MSB / MSW
print("Difference between :",diff_between,"Difference within :",diff_within)
print(f"Mean Squared Between :{MSB:.6f}")
print(f"Mean Squared Within :{MSW:.6f}")
print(f"F - Statistic : {F_stat:.6f}")
```

### Step 6 
#### p value and critical F-value

```python
alpha = 0.05
p_value = stats.f.sf(F_stat, diff_between, diff_within) # survival function
F_critical = stats.f.ppf(1-alpha, diff_between, diff_within)

print(f"p value : {p_value}")
print(f"Critical F (alpha={alpha}) : {F_critical:4f}")

if p_value < alpha :
    print("Conclusion : Reject H0 -> At least one group mean differs (Significant)")
else :
    print("Conclusion : Fail to reject H0 -> No significant difference among group means.")
```

### Optimal Summary Table

```python
print("\nANOVA table:")
print(f"{'Source':<12}{'SS':>12}{'df':>8}{'MS':>12}{'F':>10}")
print(f"{'Between':<12}{SSB:12.3f}{diff_between:8d}{MSB:12.3f}{F_stat:10.3f}")
print(f"{'Within':<12}{SSW:12.3f}{diff_within:8d}{MSW:12.3f}")
print(f"{'Total':<12}{(SSB+SSW):12.3f}{N-1:8d}")
```
