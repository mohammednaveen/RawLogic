### Hash Table From Scratch (using Chaining)

```python
# Creating a Hash Table From Scratch:
# Collission Technique used : Separate Chaining
class HashTable:
    # create hash table object
    def __init__(self,capacity=100):
        self.capacity = capacity
        self.table = [[] for _ in range(capacity)]
    # Hash Function
    def _hash(self,key):
        return hash(key) % self.capacity
    # Input an entry 
    # We are handling collissions using Seperate Chaining method
    def put(self, key, value):
        index = self._hash(key) # find the index in the table array
        bucket = self.table[index] # Select that index
        # this will be - i for the index in the table and the (k,v) is for the 
        # key value pair in the list of the list in the table
        for i, (k,_) in enumerate(bucket): 
            if k == key: # if key is already existing 
                bucket[i] = (key, value) # just update the value
                return
        bucket.append((key,value))
    # to get the value :
    def get(self,key):
        index = self._hash(key)
        bucket = self.table[index]
        for k,v in bucket:
            if k == key:
                return v
        return None

    # to remove an element from the hash table
    def remove(self,key):
        index = self._hash(key)
        bucket = self.table[index]

        for i, (k,v) in enumerate(bucket):
            if k == key:
                bucket.pop(i)
                return 
    
    # to display the whole table 
    def display(self):
        for i in range(self.capacity):
            print(i , end="")
            for key in self.table[i]:
                print("-->",key, end="")
            print()
```

### Hash Table From Scratch (with rehashing and Chaining)


```python
# creating a hash table from scratch
# collission technique = Chaining
# rehashing enabled
class HashTable:
    def __init__(self,capacity=100):
        self.capacity = capacity
        self.num_of_elements = 0
        self.table = [[] for _ in range(capacity)]

    def _hash(self,key):
        return hash(key) % self.capacity

    def put(self,key,value):
        if self.loadFactor() > 0.7:
            self.rehash()

        index = self._hash(key)
        bucket = self.table[index]

        for i , (k,v) in enumerate(bucket):
            if k == key:
                bucket[i] = ((key,value))
                return 
        bucket.append((key,value))
        self.num_of_elements += 1

    def get(self,key):
        index = self._hash(key)
        bucket = self.table[index]
        for k, v in bucket:
            if k == key:
                return v
        return None

    def remove(self,key):
        index = self._hash(key)
        bucket = self.table[index]

        for i, (k,v) in enumerate(bucket):
            if k == key:
                bucket.pop(i)
                self.num_of_elements -= 1
                return

    def display(self):
        for i in range(self.capacity):
            print(i,end="")
            for key in self.table[i]:
                print("-->",key,end="")
            print()

    def loadFactor(self):
        return self.num_of_elements/self.capacity

    def rehash(self):
        old_table = self.table
        self.capacity *= 2
        self.table = [[] for _ in range(self.capacity)]
        self.num_of_elements = 0

        for bucket in old_table:
            for key,val in bucket:
                self.put(key,val)

keys = [15, 11, 27]
hashTable = HashTable(5)

for key, val in enumerate(keys):
    hashTable.put(key,val)

hashTable.remove(11)

hashTable.display()

hashTable.put(19,9)

print("\nAfter rehashing:")
hashTable.display()
```

### Hash Table From Scratch (using Open Addressing - Linear probing) 

```python
class hashNode:
    def __init__(self,key,val):
        self.key = key
        self.val = val

class HashTable:
    def __init__(self,capacity=100):
        self.capacity = capacity
        self.table = [None for _ in range(self.capacity)]
        self.number_of_elements = 0
        self.dummy = hashNode(-1,-1)

    def hashCode(self,key):
        return hash(key) % self.capacity

    def insertNode(self,key,value):
        temp = hashNode(key, value)
        index = self.hashCode(key)

        while self.table[index] is not None and self.table[index].key != key and self.table[index] != -1:
            index = (index + 1) % self.capacity #the mod operator ensures that even if the index reaches the size of the array it stays withtin the table.
        
        if self.table[index] is None or self.table[index] == -1:
            self.number_of_elements += 1
        self.table[index] = temp

    def deleteNode(self,key):
        index = self.hashCode(key)

        while self.table[index] is not None :
            if self.table[index].key == key:
                temp = self.table[index]
                self.table[index] = self.dummy 
                self.number_of_elements -= 1
                return temp.val
            index = (index+1) % self.capacity
        
        return -1
    
    def get(self, key):
        index = self.hashCode(key)
        counter = 0

        while self.table[index] is not None:
            if counter > self.capacity:
                return -1
            if self.table[index].key == key:
                return self.table[index].val
            index = (index + 1) % self.capacity
            counter += 1
        
        return -1
    
    def display(self):
        for node in self.table:
            if node is not None and node.key != -1:
                print(f"{node.key} {node.val}")

h = HashTable()
h.insertNode(1,1)
h.insertNode(2,2)
h.insertNode(22,3)
h.display()
#print(h.deleteNode(2))
#print(h.get(2))
```

### Problems Using Hash Table


```python
def frequency(nums):
    freq = {}
    for num in nums:
        freq[num] = freq.get(num, 0) + 1
    return freq

frequency("Naveen")
```

```python
def two_sum(nums, target):
    seen = {}
    for i, num in enumerate(nums):
        diff = target - num
        if diff in seen:
            return [seen[diff], i]
        seen[num] = i
```

```python
def group_anagrams(strs):
    groups = {}
    for s in strs:
        key = ''.join(sorted(s))
        groups.setdefault(key, []).append(s)
    return list(groups.values())
```

```python
def longest_consecutive(nums):
    num_set = set(nums)
    longest = 0

    for num in num_set:
        if num - 1 not in num_set:
            length = 1
            while num + length in num_set:
                length += 1
            longest = max(longest, length)

    return longest
```
