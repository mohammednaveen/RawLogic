### Q1: Print numbers from 1 to n

```python
# this is head recursion
def printNum(num):
    if num == 1:
        return 
    printNum(num-1)
    print(num)
```

```python
printNum(10)
```

### Q2: Sum of Array Recursively

```python
def ArraySum(arr):
    n =len(arr)
    if n == 0:
        return 0
    return arr[0] + ArraySum(arr[1:])
```

```python
sum = ArraySum([1,2,3,4,5])
sum
```

### Q3: Check if a string is palindrome recursively

```python
def Palindrome(str):
    left = 0
    right = len(str)-1
    if left >= right:
        return True
    if str[left] != str[right]:
        return False
    else:
        Palindrome(str[left+1:right])
    return True
```

```python
def Palindrome3(s):
    if len(s) <= 1:
        return True
    if s[0] != s[-1]:
        return False
    return Palindrome3(s[1:-1])

print(Palindrome3("ababa"),"third function")
```

```python
def Palindrome2(s,left,right):
    if left >= right:
        return True
    if s[left] != s[right]:
        return False
    
    return Palindrome2(s,left+1,right-1)
```

```python
res = Palindrome("malayalam")
print(res, "first function")
res2 = Palindrome2("malayalam",0,len("malayalam")-1)
print(res2,"second function")
```

### Q4: Power function (xⁿ)

```python
def PowerFunction(x,n):
    if n == 0:
        return 1
    return x * PowerFunction(x,n-1)
# Time complexity : O(n)
# Space Complexity : O(n)
```

```python
power = PowerFunction(2,3)
power
```

```python
# Divide and Conquer
def Power(x,n):
    if n == 0:
        return 1
    half = Power(x,n//2)

    if n % 2 == 0:
        return half * half
    else:
        return x * half *half
# Time Complexity : O(log n)
# Space Complexity : O(log n)
```

```python
power = Power(2,3)
power
```

### Q5: Factorial 

```python
def factorial(n):
    if n == 1:
        return 1
    return n * factorial(n-1)
```

```python
res = factorial(5)
res
```

### Q6: Bit Strings

```python
def bitStrings(n):
    if n == 0: return []
    if n == 1: return["0","1"]
    return [
        bitstring + digit for digit in bitStrings(1)
        for bitstring in bitStrings(n-1)
    ]
print(bitStrings(3))
```

### Q7: Check if Power of Two

```python
def powerOfTwo(n):
    if n < 1:
        return False
    while n > 1:
        if n % 2 != 0:
            return False
        n = n/2
    return True
```

### Q8: Count digits of a number:

```python
def CountNum(num):
    num = abs(num)
    if num <= 0:
        return 0
    return 1 + CountNum(num//10)
```
