```python
# Find max element:
def findMax(arr):
    max_var = arr[0]
    for i in arr:
        if max_var <  i:
            max_var = i
    return max_var

findMax([1,2,6,3,2])
```

```python
# Find min element:
def findMin(arr):
    min_var = arr[0]
    for i in arr:
        if min_var > i:
            min_var = i
    return min_var

findMin([2,4,1,4,5])
```

```python
# Count frequency of each element:
def frequency(arr):
    freq = {}
    for num in arr:
        freq[num] = freq.get(num,0) + 1
    return freq

frequency([1,1,2,34,4,2,4,2,5,7,3,2,3,3])
```

```python
# Check if array is Sorted:
def is_sorted(arr):
    for i in range(len(arr)-1):
        if arr[i] > arr[i+1]:
            return False
    return True

is_sorted([1,2,3,5,6])
```

```python
# Reverse array in place:
def reverse_array(arr):
    left = 0
    right = len(arr)-1 
    while left < right:
        arr[left],arr[right] = arr[right],arr[left]
        left += 1
        right -= 1

    return arr

reverse_array([1,2,3,4,5])
```

```python
# Rotate the first element of the array:
def rotate_by_one(arr):
    last = arr[-1]

    for i in range(len(arr)-1, 0, -1):
        arr[i] = arr[i-1]

    arr[0] = last
    return arr

rotate_by_one([1,2,3,4,5])
```

```python
# Rotate the array k times:
def rotate_k(arr,k):
    n = len(arr)
    k = k%n
    def reverse(l,r):
        while l<r:
            arr[l],arr[r] = arr[r],arr[l]
            l += 1
            r -= 1
    reverse(0,n-1)
    reverse(0,k-1)
    reverse(k,n-1)
    return arr

rotate_k([1,2,3,4,5,6,7],2)
```
