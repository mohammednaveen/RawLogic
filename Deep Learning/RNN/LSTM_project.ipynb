```python
import tensorflow as tf
from tensorflow.keras.preprocessing.text import Tokenizer
```

```python
tokenizer = Tokenizer()
```

```python
tokenizer.fit_on_texts([text])
```

```python
vocab_size = len(tokenizer.word_index) + 1
tokenizer.word_index
```

```python
input_sequences = []
for sentence in text.split('\n'):
  # print(sentence)
  tokenized_sentence = tokenizer.texts_to_sequences([sentence])[0]

  for i in range(1, len(tokenized_sentence)):
    input_sequences.append(tokenized_sentence[:i+1])
```

```python
input_sequences
```

```python
max_length = max([len(x) for x in input_sequences])
```

```python
from tensorflow.keras.preprocessing.sequence import pad_sequences
padded_input_sequences = pad_sequences(input_sequences, maxlen=max_length, padding="pre")
```

```python
padded_input_sequences
```

```python
X = padded_input_sequences[:,:-1]
```

```python
y = padded_input_sequences[:,-1]
```

```python
X.shape, y.shape
```

```python
# taking how many words are there - len of vocabulary
len(tokenizer.word_index) # 333 unique words
```

```python
from tensorflow.keras.utils import to_categorical # converting labels (unique word index) to identifyable vectors for classification
y = to_categorical(y, num_classes=vocab_size) # 334 cuz tokenizer starts from 1 so the last 333th word will be missed if we give 333 so we gave +1 i.e 334 to include that word in too.
```

```python
y
```

```python
y.shape
```

```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding, LSTM, Dense
```

```python
model = Sequential()
model.add(Embedding(vocab_size, 100))
model.add(LSTM(180)) # 180 nodes
model.add(Dense(vocab_size, activation="softmax"))
```

```python
model.compile(
    loss = "categorical_crossentropy",
    optimizer = "adam",
    metrics = ["accuracy"]
)
```

```python
model.summary()
```

```python
model.fit(X,y,epochs=100)
```

```python
import numpy as np
import time
test_text = "Why"

for i in range(5):
  # tokenize
  token_text = tokenizer.texts_to_sequences([test_text])[0]
  # padding
  padded_text = pad_sequences([token_text], maxlen=X.shape[1], padding="pre")
  # predict
  pos = np.argmax(model.predict(padded_text))

  for word, index in tokenizer.word_index.items():
    if index == pos:
      test_text = test_text + " " + word
      print(test_text)
      time.sleep(2)
```
