### Neural Network from scratch using NumPy

```python
import numpy as np
```

```python
# data
X = np.array([
    [0.8, 0.6],
    [0.3, 0.9],
    [0.7, 0.4],
    [0.2, 0.5]
]) # (4x2)

y = np.array([
    [1],
    [1],
    [0],
    [0]
]) # (1x4)
```

```python
# initialize the parameters
w1 = np.random.randn(2,2)
b1 = np.random.randn(1,2)

w2 = np.random.randn(2,1)
b2 = np.random.randn(1,1)
```

```python
# Activation Functions
def relu(z):
    return np.maximum(0,z)

def sigmoid(z):
    return (1 / (1 + np.exp(-z)) )
```

```python
# Training :
m = X.shape[0]
epochs = 1000
lr = 0.1

for epoch in range(epochs):

    # Forward Propagation
    z1 = X @ w1 + b1 # (4x2)
    a1 = relu(z1) # (4x2)

    z2 = a1 @ w2 + b2 # (4x1)
    y_pred = sigmoid(z2)

    # Loss Calculation
    L = - np.mean(
        y * np.log(y_pred) + (1-y) * np.log(1-y_pred)
    )

    
    # Backpropagation
    dz2 = (y_pred - y) / m # (4x1)
    # dL/dz2 = dL/dy_pred . dy_pred/dz2 -> which in turn gives us (y_pred - y)
    # m is there since we have multiple features
    dw2 = a1.T @ dz2 # (2x1)
    # the calculation is done in this order to give us the added paths of backpropagation
    # dL/dw2 = dL/y_pred . dy_pred/dz2 . dz2/dw2 -> which in turn gives us dz2 * a1 (output of the hidden layer in the equation z2 = a1 * w2 + b2)
    db2 = np.sum(dz2, axis=0, keepdims=True) # added loss paths (1x1)
    # dL/db2 = dL/y_pred . dy_pred/dz2 . dz2/db2 -> which in turn gives us dz2 * 1 (output of the hidden layer in the equation z2 = a1 * w2 + b2)

    da1 = dz2 @ w2.T # (4x2)
    # dL/da1 = dL/y_pred . dy_pred/dz2 . dz2/da1 -> which in turn gives us dz2 * w2 (output of the hidden layer in the equation z2 = a1 * w2 + b2)
    dz1 = da1 * (z1 > 0) # (4x2)
    # dL/dz1 = dL/y_pred . dy_pred/dz2 . dz2/da1 . da1/dz1 -> which in turn gives us da1 * (z1 > 0) (derivative of the relu function if > 0 -> 1, if < 0 -> 0)
    # (Z1 > 0) returns boolean matrix which allows us to do element wise multiplication since that is the thing that shall be done.

    dw1 = X.T @ dz1 # (2x2)
    # the calculation is done in this order to give us the added paths of backpropagation
    # dL/dw1 = dL/y_pred . dy_pred/dz2 . dz2/da1 . da1/dz1 . dz1/dw1 -> which in turn gives us dz1 * X (from the equation z1 = X * w1 + b1)
    db1 = np.sum(dz1, axis=0, keepdims=True) # (1x2)
    # addition because we have to add all the paths leading to b1
    # dL/db1 = dL/y_pred . dy_pred/dz2 . dz2/da1 . da1/dz1 . dz1/db1 -> which in turn gives us dz1 * 1 (from the equation z1 = X * w1 + b1)


    # Gradient Descent
    w2 -= lr * dw2
    b2 -= lr * db2

    w1 -= lr * dw1
    b1 -= lr * db1

    if epoch % 100 == 0:
        print(f"Epoch {epoch}, Loss: {L:.4f}")
```
