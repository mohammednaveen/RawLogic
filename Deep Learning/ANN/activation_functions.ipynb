## Activation Functions for Neural Networks

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# listing down all the activation functions

# Sigmoid: σ(x) = 1 / (1 + e^(-x))
def sigmoid(x):
    return 1 / (1 + np.exp(-x))

# ReLU: ReLU(x) = max(0, x)
def relu(x):
    return np.maximum(0,x)

# Tanh: tanh(x) = (e^x - e^(-x)) / (e^x + e^(-x))
def tanh(x):
    return np.tanh(x)

# Leaky ReLU:
# f(x) = x       if x > 0
#        0.01x   if x <= 0
def leaky_relu(x):
    return np.where(
        x > 0, 
        x,
        0.01 * x
    )

# Softmax:
# softmax(x_i) = e^(x_i) / Σ e^(x_j)
def softmax(x):
    exp_x = np.exp(x - max(x))
    return exp_x / np.sum(exp_x)
```

```python
x = np.linspace(-5, 5, 500)
```

```python
plt.plot(x, sigmoid(x))

plt.title(r"Sigmoid: $\sigma(x)=\frac{1}{1+e^{-x}}$")
plt.xlabel("x")
plt.ylabel("σ(x)")
plt.grid(True)

plt.show()
```

```python
plt.plot(x, tanh(x))

plt.title(r"Tanh: $\tanh(x)=\frac{e^x-e^{-x}}{e^x+e^{-x}}$")
plt.xlabel("x")
plt.ylabel("tanh(x)")
plt.grid(True)

plt.show()
```

```python
plt.plot(x, relu(x))

plt.title(r"ReLU: $f(x)=\max(0,x)$")
plt.xlabel("x")
plt.ylabel("ReLU(x)")
plt.grid(True)

plt.show()
```

```python
plt.plot(x, leaky_relu(x))

plt.title(r"Leaky ReLU: $f(x)=\max(0.01x,x)$")
plt.xlabel("x")
plt.ylabel("Leaky ReLU(x)")
plt.grid(True)

plt.show()
```

```python
scores = np.array([2.0, 1.0, 0.1])

probabilities = softmax(scores)

print("Scores:", scores)
print("Softmax:", probabilities)
print("Sum:", np.sum(probabilities))
```

```python
scores = np.linspace(-5, 5, 200)

p1 = []
p2 = []
p3 = []

for s in scores:
    probs = softmax(np.array([s, 1, 0]))
    p1.append(probs[0])
    p2.append(probs[1])
    p3.append(probs[2])

plt.figure(figsize=(7, 4))

plt.plot(scores, p1, label="Class 1")
plt.plot(scores, p2, label="Class 2")
plt.plot(scores, p3, label="Class 3")

plt.title("Softmax Output")
plt.xlabel("Class 1 Score")
plt.ylabel("Probability")
plt.legend()
plt.grid(True)

plt.show()
```
