### Neural Network written properly

```python
import numpy as np
```

```python
class NeuralNetwork:

    def __init__(self, input_size=2, hidden_size=2, output_size=1, lr=0.1):
        self.lr = lr

        # parameters
        self.w1 = np.random.randn(input_size, hidden_size)
        self.b1 = np.zeros((1, hidden_size))

        self.w2 = np.random.randn(hidden_size,output_size)
        self.b2 = np.zeros((1, output_size))

    def relu(self, z):
        return np.maximum(0,z)

    def sigmoid(self, z):
        return 1 / (1 + np.exp(-z))
    
    def forward(self, X):
        self.z1 = X @ self.w1 + self.b1
        self.a1 = self.relu(self.z1)

        self.z2 = self.a1 @ self.w2 + self.b2
        self.y_pred = self.sigmoid(self.z2)

        return self.y_pred

    def loss(self, y, y_pred):
        return -np.mean(
            y * np.log(y_pred) + (1-y) * np.log(1-y_pred)
        )

    def backward(self, X, y):
        m = X.shape[0]

        dz2 = (self.y_pred - y) / m

        self.dw2 = self.a1.T @ dz2
        self.db2 = np.sum(dz2, axis=0 ,keepdims=True)

        da1 = dz2 @ self.w2.T
        dz1 = da1 * (self.z1 > 0)

        self.dw1 = X.T @ dz1
        self.db1 = np.sum(dz1, axis=0, keepdims=True)

    def update(self):
        self.w2 = self.w2 - self.lr*self.dw2
        self.b2 = self.b2 - self.lr*self.db2

        self.w1 = self.w1 - self.lr*self.dw1
        self.b1 = self.b1 - self.lr*self.db1

    def train(self, X, y, epochs=1000):

        for epoch in range(epochs):
            
            # Forward propagation
            y_pred = self.forward(X)

            # Loss
            L = self.loss(y,y_pred)

            # Backpropagate
            self.backward(X,y)

            # Gradient Descent 
            self.update()

            if epoch % 100 == 0:
                print(f"Epoch : {epoch}, Loss : {L}")

    def prediction(self,X):
        prob = self.forward(X)
        
        return (prob >= 0.5).astype(int)
```

```python
X = np.array([
    [0.8, 0.6],
    [0.3, 0.9],
    [0.7, 0.4],
    [0.2, 0.5]
])

y = np.array([
    [1],
    [1],
    [0],
    [0]
])

model = NeuralNetwork(lr=0.1)

model.train(X, y, epochs=1000)
```

```python
new_student = np.array([[0.9, 0.8]])

print(model.prediction(new_student))
```
